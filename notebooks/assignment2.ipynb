```python
import pandas as pd
from pathlib import Path

# Exact location of your dataset
csv_path = (
    Path.home()
    / "OneDrive - Takeda"
    / "Desktop"
    / "manufacturing_defect_classifier"
    / "data"
    / "manufacturing_defect_dataset.csv"
)

# Load the dataset
df = pd.read_csv(csv_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
# 1. Check the possible target values
print("Unique DefectStatus values:")
print(sorted(df["DefectStatus"].unique()))

# 2. Count each class
print("\nNumber of records per class:")
print(df["DefectStatus"].value_counts().sort_index())

# 3. Check the percentage of each class
print("\nClass distribution (%):")
print(
    (df["DefectStatus"].value_counts(normalize=True) * 100)
    .sort_index()
    .round(2)
)

# 4. Visualize the class balance
df["DefectStatus"].value_counts().sort_index().plot(
    kind="bar",
    title="Distribution of Defect Status",
    xlabel="Defect Status",
    ylabel="Number of records",
    rot=0
);
```

```python
# Compare the average feature values across the two target classes
class_means = df.groupby("DefectStatus").mean(numeric_only=True).T

class_means["AbsoluteDifference"] = (
    class_means[0] - class_means[1]
).abs()

display(
    class_means.sort_values(
        "AbsoluteDifference",
        ascending=False
    ).round(3)
)
```

```python
# Correlation of each numerical feature with the target
correlations = (
    df.corr(numeric_only=True)["DefectStatus"]
    .drop("DefectStatus")
    .sort_values(key=abs, ascending=False)
)

display(correlations.to_frame("Correlation with DefectStatus").round(3))
```

```python
# Inspect the target classes against key quality-related variables

for col in ["DefectRate", "QualityScore", "MaintenanceHours"]:
    print(f"\n--- {col} by DefectStatus ---")

    print(
        df.groupby("DefectStatus")[col]
        .agg(["min", "median", "max", "mean", "std"])
        .round(3)
    )
```

```python
# Examine ranges for each target class

for col in ["DefectRate", "QualityScore"]:
    print(f"\n--- {col}: class ranges ---")

    print(
        df.groupby("DefectStatus")[col]
        .agg(["min", "max"])
        .round(3)
    )
```

```python
from sklearn.model_selection import train_test_split

# Separate input features and target
X = df.drop(columns=["DefectStatus"])
y = df["DefectStatus"]

# Split: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Build the Logistic Regression pipeline
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Train on training data only
logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# Predict class labels and probabilities
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

# Calculate metrics
logistic_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_test, y_pred),
    "AUC": roc_auc_score(y_test, y_prob),
    "Precision": precision_score(y_test, y_pred, zero_division=0),
    "Recall": recall_score(y_test, y_pred, zero_division=0),
    "F1 Score": f1_score(y_test, y_pred, zero_division=0),
    "MCC": matthews_corrcoef(y_test, y_pred)
}

for metric, value in logistic_results.items():
    if isinstance(value, (int, float)):
        print(f"{metric}: {value:.4f}")
    else:
        print(f"{metric}: {value}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(
    max_depth=6,
    min_samples_leaf=10,
    class_weight="balanced",
    random_state=42
)

tree_model.fit(X_train, y_train)

print("Decision Tree trained successfully!")
```

```python
# Predict class labels and probabilities
tree_pred = tree_model.predict(X_test)
tree_prob = tree_model.predict_proba(X_test)[:, 1]

# Calculate all six required metrics
tree_results = {
    "Model": "Decision Tree",
    "Accuracy": accuracy_score(y_test, tree_pred),
    "AUC": roc_auc_score(y_test, tree_prob),
    "Precision": precision_score(y_test, tree_pred, zero_division=0),
    "Recall": recall_score(y_test, tree_pred, zero_division=0),
    "F1 Score": f1_score(y_test, tree_pred, zero_division=0),
    "MCC": matthews_corrcoef(y_test, tree_pred)
}

for metric, value in tree_results.items():
    if isinstance(value, (int, float)):
        print(f"{metric}: {value:.4f}")
    else:
        print(f"{metric}: {value}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, tree_pred))

print("\nClassification Report:")
print(classification_report(y_test, tree_pred, zero_division=0))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)

print("KNN trained successfully!")
```

```python
from sklearn.model_selection import train_test_split

excluded_features = ["DefectRate", "QualityScore"]

X_reduced = df.drop(
    columns=["DefectStatus"] + excluded_features
)
y = df["DefectStatus"]

X_train_reduced, X_test_reduced, y_train_reduced, y_test_reduced = (
    train_test_split(
        X_reduced,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

print("Original feature count:", X.shape[1])
print("Reduced feature count:", X_reduced.shape[1])
print("Training shape:", X_train_reduced.shape)
print("Test shape:", X_test_reduced.shape)
```

```python
from sklearn.base import clone
from sklearn.metrics import (
    accuracy_score, roc_auc_score, precision_score,
    recall_score, f1_score, matthews_corrcoef
)
import pandas as pd

# Reuse the original model configurations
models_to_compare = {
    "Logistic Regression": logistic_model,
    "Decision Tree": tree_model,
    "KNN": knn_model,
    "Naive Bayes": nb_model,
    "SVM": svm_model
}

reduced_results = []
reduced_models = {}

for name, original_model in models_to_compare.items():
    model = clone(original_model)

    model.fit(X_train_reduced, y_train_reduced)

    predictions = model.predict(X_test_reduced)
    probabilities = model.predict_proba(X_test_reduced)[:, 1]

    reduced_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_reduced, predictions),
        "AUC": roc_auc_score(y_test_reduced, probabilities),
        "Precision": precision_score(
            y_test_reduced, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test_reduced, predictions, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test_reduced, predictions, zero_division=0
        ),
        "MCC": matthews_corrcoef(y_test_reduced, predictions)
    })

    reduced_models[name] = model

reduced_results_df = pd.DataFrame(reduced_results).set_index("Model")

display(reduced_results_df.round(4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

for name, model in reduced_models.items():
    predictions = model.predict(X_test_reduced)

    print(f"\n{'=' * 45}")
    print(name)
    print("Confusion Matrix:")
    print(confusion_matrix(y_test_reduced, predictions))
    print("\nClassification Report:")
    print(classification_report(
        y_test_reduced,
        predictions,
        zero_division=0
    ))
```

```python

app_source = r'''
from pathlib import Path

import joblib
import pandas as pd
import streamlit as st

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report,
)

st.set_page_config(
    page_title="Manufacturing Defect Classifier",
    page_icon="🏭",
    layout="wide",
)

BASE_DIR = Path(__file__).resolve().parent
MODEL_DIR = BASE_DIR / "model"

MODEL_FILES = {
    "Logistic Regression": "logistic_regression.joblib",
    "Decision Tree": "decision_tree.joblib",
    "K-Nearest Neighbors (KNN)": "knn.joblib",
    "Gaussian Naive Bayes": "naive_bayes.joblib",
    "Support Vector Machine (SVM)": "svm.joblib",
}

FEATURE_COLUMNS = [
    "ProductionVolume",
    "ProductionCost",
    "SupplierQuality",
    "DeliveryDelay",
    "DefectRate",
    "QualityScore",
    "MaintenanceHours",
    "DowntimePercentage",
    "InventoryTurnover",
    "StockoutRate",
    "WorkerProductivity",
    "SafetyIncidents",
    "EnergyConsumption",
    "EnergyEfficiency",
    "AdditiveProcessTime",
    "AdditiveMaterialCost",
]
TARGET_COLUMN = "DefectStatus"


@st.cache_resource
def load_model(model_path):
    return joblib.load(model_path)


st.title("🏭 Manufacturing Defect Classification")
st.write(
    "Evaluate five trained machine-learning classifiers "
    "using labelled test data."
)
st.caption(
    "The app loads saved models and evaluates the uploaded CSV. "
    "It does not retrain models."
)

st.info(
    "Upload a CSV containing all 16 input features and "
    "the DefectStatus target column."
)

uploaded_file = st.file_uploader(
    "Upload labelled test data",
    type=["csv"],
)

selected_model_name = st.selectbox(
    "Select a classification model",
    list(MODEL_FILES.keys()),
)

if uploaded_file is None:
    st.caption("Use test_data.csv from your project folder to begin.")
else:
    try:
        test_df = pd.read_csv(uploaded_file)

        required_columns = FEATURE_COLUMNS + [TARGET_COLUMN]
        missing_columns = [
            column for column in required_columns
            if column not in test_df.columns
        ]

        if missing_columns:
            st.error(
                "Missing required columns: "
                + ", ".join(missing_columns)
            )
            st.stop()

        if test_df.empty:
            st.error("The uploaded CSV contains no rows.")
            st.stop()

        # Select only required columns in the expected order.
        X_uploaded = test_df[FEATURE_COLUMNS].copy()
        y_true = test_df[TARGET_COLUMN]

        # Check for missing and non-numeric feature values.
        X_uploaded = X_uploaded.apply(
            pd.to_numeric, errors="coerce"
        )
        y_numeric = pd.to_numeric(y_true, errors="coerce")

        if X_uploaded.isnull().any().any() or y_numeric.isnull().any():
            st.error(
                "Missing or non-numeric values were found "
                "in the required columns. Please clean the CSV."
            )
            st.stop()

        if not y_numeric.isin([0, 1]).all():
            st.error(
                "DefectStatus must contain only the labels 0 and 1."
            )
            st.stop()

        y_true = y_numeric.astype(int)

        if y_true.nunique() != 2:
            st.error(
                "AUC requires both target classes (0 and 1) "
                "to be present in the uploaded test data."
            )
            st.stop()

        model_path = MODEL_DIR / MODEL_FILES[selected_model_name]

        if not model_path.exists():
            st.error(
                f"Saved model not found: {model_path.name}. "
                "Check the model folder."
            )
            st.stop()

        model = load_model(str(model_path))

        y_pred = model.predict(X_uploaded)
        probabilities = model.predict_proba(X_uploaded)

        # Locate class 1 explicitly for the AUC calculation.
        class_labels = list(model.classes_)
        positive_class_index = class_labels.index(1)
        y_prob = probabilities[:, positive_class_index]

        metrics = {
            "Accuracy": accuracy_score(y_true, y_pred),
            "AUC": roc_auc_score(y_true, y_prob),
            "Precision": precision_score(
                y_true, y_pred, zero_division=0
            ),
            "Recall": recall_score(
                y_true, y_pred, zero_division=0
            ),
            "F1 Score": f1_score(
                y_true, y_pred, zero_division=0
            ),
            "MCC": matthews_corrcoef(y_true, y_pred),
        }

        st.subheader(f"Evaluation: {selected_model_name}")
        st.write(f"**Test records:** {len(test_df):,}")
        st.caption(
            "Precision, recall and F1 use class 1 as the positive class."
        )

        metric_columns = st.columns(3)
        for column, metric_name in zip(
            metric_columns, ["Accuracy", "AUC", "Precision"]
        ):
            column.metric(metric_name, f"{metrics[metric_name]:.4f}")

        metric_columns = st.columns(3)
        for column, metric_name in zip(
            metric_columns, ["Recall", "F1 Score", "MCC"]
        ):
            column.metric(metric_name, f"{metrics[metric_name]:.4f}")

        st.subheader("Confusion Matrix")
        cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
        cm_df = pd.DataFrame(
            cm,
            index=["Actual 0", "Actual 1"],
            columns=["Predicted 0", "Predicted 1"],
        )
        st.dataframe(cm_df, use_container_width=True)

        st.subheader("Classification Report")
        report = classification_report(
            y_true,
            y_pred,
            labels=[0, 1],
            output_dict=True,
            zero_division=0,
        )
        st.dataframe(
            pd.DataFrame(report).transpose().round(4),
            use_container_width=True,
        )

    except Exception as exc:
        st.error(
            "The file could not be evaluated. Check its format "
            f"and the saved models. Details: {exc}"
        )
'''

app_path.write_text(app_source, encoding="utf-8")
print("Streamlit app saved:", app_path)
print("File size:", app_path.stat().st_size, "bytes")
```
